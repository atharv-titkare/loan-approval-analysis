# EDA

### Univariate Analysis

```python
sns.countplot(x = df.Gender)
```

```python
Categorical = []

for i in df:
    if len(df[i].unique())<=20:
        Categorical.append(i)
```

```python
Categorical
```

```python
Countinous = []

for i in df:
    if len(df[i].unique())>=20:
        Countinous.append(i)
```

```python
Countinous
```

```python
df_cat = df[Categorical]
df_cat
```

```python
df_cont = df[Countinous]
df_cont
```

```python
df_cont = df_cont.drop(columns = 'Loan_ID')
df_cont
```

```python
df_cat.columns
```

```python
plotnumber = 1

plt.figure(figsize=(10, 10))
for i in df_cat:
    plt.subplot(3, 3, plotnumber)
    sns.countplot(x= df_cat[i])
    plt.xticks(rotation=45)
    plotnumber=plotnumber+1
plt.tight_layout()
```

Insights

1.Gender : Male applicants are higher than female applicants.The Dataset is male-dominated
2.Married : Most applicants are marrried.
3.Dependents : Most applicants have 0 dependents.The number of applicants decreases as dependents increase.
4.Education : Most applicants are Graduates.Graduates applicants make up the majority of loan applications.
5.Self Employed : Very few applicants are self-employed.
6.Loan Amount Term : The 360-months(30 years)loan term is by far the most common
7.Credit History : Applicants with a good credit history are muh more common and generally more likely to get loan approval
8.property Area : Semiurban area contribute the largest share of loan applications.
9.Loan Status : The dataset is slightly imbalanced because approved loans are more frequent than rejected loans         

```python
plotnumber = 1

plt.figure(figsize=(10, 5))
for i in df_cont:
    plt.subplot(1, 3, plotnumber)
    sns.histplot(x= df_cont[i])
    plt.xticks(rotation=45)
    plotnumber=plotnumber+1
plt.tight_layout()
```

### Bivariate Analysis

```python
plotnumber = 1

plt.figure(figsize=(10, 10))
for i in df_cat:
    plt.subplot(3, 3, plotnumber)
    sns.countplot(x= df_cat[i], hue = df.Loan_Status)
    plt.xticks(rotation=45)
    plotnumber=plotnumber+1
plt.tight_layout()
```

Insights

```python
plotnumber = 1

plt.figure(figsize=(10, 5))
for i in df_cont:
    plt.subplot(1, 3, plotnumber)
    sns.histplot(x= df_cont[i],hue = df.Loan_Status, kde = True)
    plt.xticks(rotation=45)
    plotnumber=plotnumber+1
plt.tight_layout()
```

### Data Cleaning

#### Handling with null values

```python
df.isnull().sum()
```

```python
df.loc[df.Gender.isnull(),['Gender']]
```

```python
df.loc[df.Gender.isnull(),['Gender']] = 'Male'

# Or
# df.Gender = df.Gender.fillna(value = 'Male')
```

```python
df.isnull().sum()
```

```python
df.Married.value_counts()
```

```python
df.loc[df.Married.isnull(),'Married']='Yes'
```

```python
df.Dependents.value_counts()
```

```python
df.loc[df.Dependents.isnull(),'Dependents'] = '0'
```

```python
df.Self_Employed.value_counts()
```

```python
df.loc[df.Self_Employed.isnull(),'Self_Employed'] = 'No'
```

```python
df.loc[df.LoanAmount.isnull(),'LoanAmount']
```

```python
np.median(df.LoanAmount.dropna())
```

```python
df.loc[df.LoanAmount.isnull(),'LoanAmount'] = np.median(df.LoanAmount.dropna())
```

```python
df.Loan_Amount_Term.value_counts()
```

```python
df.loc[df.Loan_Amount_Term.isnull(),'Loan_Amount_Term'] = 360.0
```

```python
df.Credit_History.value_counts()
```

```python
df.loc[df.Credit_History.isnull(),'Credit_History'] = 1.0
```
